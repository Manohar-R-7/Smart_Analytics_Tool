```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.info()
```

```python
missing = df.isnull().sum()

print(missing)
```

```python
df.describe()
```

```python
wins = df['winner'].value_counts().reset_index()

wins.columns = ['Team', 'Wins']

fig = px.bar(
    wins,
    x='Team',
    y='Wins',
    title='Team Wins Analysis'
)

fig.show()
```

```python
toss = df['toss_winner'].value_counts().reset_index()

toss.columns = ['Team', 'Count']

fig = px.pie(
    toss,
    names='Team',
    values='Count',
    title='Toss Winners Distribution'
)

fig.show()
```

```python
matches = df.groupby('season').size().reset_index(name='Matches')

fig = px.line(
    matches,
    x='season',
    y='Matches',
    title='Matches Per Season'
)

fig.show()
```
